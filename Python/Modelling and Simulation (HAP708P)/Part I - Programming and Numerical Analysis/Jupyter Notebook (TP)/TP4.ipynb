<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>1) Basic commands of the module Pyplot from the package Matplotlib</i></span>

Matplotlib is a package to make figures. The most relevant module for us is <span style="color: #AB68B2;"><i>Pyplot</i></span>. We list below the most relevant functions. For more details, consult the lectures notes <strong>§III.1</strong>, or look at the documentation: https://matplotlib.org/3.5.3/api/_as_gen/matplotlib.pyplot.html.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>a) The function <tt>figure</tt></i></span>

The function <tt>figure</tt> is used to initialize the graph. Without this statement, Python will initialize a new figure with default characteristics. So using this function allows you to start a clean environment, possibly tailored to your needs. 

It is possible to pass a window name (type <tt>str</tt>) or a number as an argument. In this case, if the same program is run several times without closing the window, the graph is plotted as many times with different colours.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>b) The function <tt>plot</tt></i></span>

The function <tt>plot</tt> is probably the most used function in scientific visualization. It allows you to plot two lists, or two NumPy arrays, which contain the abscissae and ordinates.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>c) The function <tt>show</tt></i></span>

The function <tt>show</tt> requests the display of all open figures. It is used once all the operations for constructing the figures have been completed. Thus, many plotting actions can be accumulated before displaying.

The function <tt>show</tt> does not close the figure. It is still possible to add elements to it later until the <tt>close</tt> instruction (if you do not manually close the figure). <strong> This may not work in Jupyter notebooks which may have an "auto-clearing" behaviour after <tt>show</tt><strong> (but we will not discuss this in detail).

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>d) The function <tt>close</tt></i></span>

The function <tt>close</tt> is never used in interactive mode (<i>i.e.,</i> when you directly pass commands after running <tt>python</tt> or <tt>python3</tt> in a terminal) as it destroys the window. 

In programming, it is a good habit to place a <tt>close</tt> before the call to <tt>figure</tt> in order to recreate a new figure and not to take the risk of rewriting in an already existing figure of the same name.

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>3) Drawing several curves on the same graph </i></span>

The range of the two axes is adjusted automatically. 

You can <span style="color: #AB68B2;"><i>label</i></span> the different curves and show the legend of the plot.

```python
plt.figure("Plotting three functions")
x = np.linspace(0, 2 * np.pi, 50)
plt.plot(x, np.cos(x), label="cos(x)")
x = np.linspace(- np.pi, np.pi, 50)
plt.plot(x, np.sin(x), label="sin(x)")
x = np.linspace(-5, 5, 50)
plt.plot(x, 1 / (1 + np.exp(-x)), label="1/(1+e^{-x})")
plt.legend()  # By default, Python tries to find the best location of the legend to avoid overlap with the curves.
plt.show()
```

```python
plt.figure("Plotting three functions")
x = np.linspace(-np.pi, 2 * np.pi, 50)
xx = x.reshape((x.size, 1))  # turns the line of x into a column
y = np.hstack((np.cos(xx), np.sin(xx)))  # two columns, one with cos, one with sin ?
plt.plot(x, y, label=['cos(x)', 'sin(x)'])  # plot x, y, but since y is carrying two columns it plots two curves ?
x = np.linspace(-5, 5, 50)
plt.plot(x, 1 / (1 + np.exp(-x)), label="1/(1+e^{-x})")
plt.legend()
plt.show()
```

```python
plt.figure("Plotting three functions")
x = np.linspace(0, 2 * np.pi, 50)
plt.plot(x, np.cos(x), label="cos(x)")
x = np.linspace(- np.pi, np.pi, 50)
plt.plot(x, np.sin(x), label="sin(x)")
x = np.linspace(-5, 5, 50)
plt.plot(x, 1 / (1 + np.exp(-x)), label="1/(1+e^{-x})")
plt.legend(loc=0)  # loc = 0 means that Matplotlib tries to find the best location, but you can try changing the value of loc (integers
# between 0 and 10 to see how the legend moves).
plt.show()
```

```python
plt.figure("Plotting three functions")
x = np.linspace(0, 2 * np.pi, 50)
plt.plot(x, np.cos(x), label=r"$\cos(x)$")  # To write the legend with LaTeX, you have to put an 'r' before the string, and then you can
# use LaTeX commands in the string.
x = np.linspace(- np.pi, np.pi, 50)
plt.plot(x, np.sin(x), label=r"$\sin(x)$")
x = np.linspace(-5, 5, 50)
plt.plot(x, 1 / (1 + np.exp(-x)), label=r"$\dfrac{1}{1+e^{-x}}$")
plt.legend()
plt.show()
```

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>4) Defining axes, grid and labels </i></span>

A readable figure should include titles on axes and the values it displays should be readable.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>a) The range</i></span>

The function <tt>axis([xmin, xmax, ymin, ymax])</tt> defines the <span style="color: #AB68B2;"><i>extent of the two axes</i></span>.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>b) The grid </i></span>

For the figure to be more readable you can add <span style="color: #AB68B2;"><i>a grid</i></span> with the function <tt>grid</tt>.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>c) Title of the axes</i></span>

The functions <tt>xlabel('title of the x-axis')</tt> and <tt>ylabel('title of the y-axis')</tt> allow you to give <span style="color: #AB68B2;"><i>titles to the axes</i></span>.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>d) Title of the graph</i></span>

To add a <span style="color: #AB68B2;"><i>title to the entire graph</i></span>, you can use the function <tt>title</tt>.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>e) Ticks on the axes</i></span>

You can specify <span style="color: #AB68B2;"><i>the ticks</i></span> on the x-axis and on the y-axis with the functions <tt>xticks(list_of_ticks, list_of_tick_labels)</tt> and <tt>yticks(list_of_ticks, list_of_tick_labels)</tt>.

<span style="color: #1BBAC6; font-size: 16px; margin-left: 4em;"><i>f) Changing the scale of axes</i></span>

It is sometimes useful to show data in <span style="color: #AB68B2;"><i>logarithmic scale rather than in linear scale</i></span>. To switch one axis from a linear scale (default) to a logarithmic scale, you can use the function <tt>xscale('log')</tt> or <tt>yscale('log')</tt>. You can then switch back to a linear scale using <tt>xscale('lin')</tt> and <tt>yscale('lin')</tt>.

```python
plt.figure()
x = np.linspace(0, 2 * np.pi, 50)
plt.subplot(2, 1, 1)  # stack multiple graphs on the same window (nrow, ncolumn, index of the current thing being plotted)
plt.plot(x, np.cos(x))
plt.ylabel(r'$\cos(x)$')
plt.subplot(2, 1, 2)  # same grid of subplot, but index 2 (bottom one)
plt.plot(x, np.sin(x))
plt.ylabel(r'$\sin(x)$')
plt.xlabel('x')
plt.suptitle('Trigonometric functions')  # title common to both graphs
plt.show()
```

```python
plt.figure()
x = np.linspace(0, 2 * np.pi, 50)
plt.subplot2grid((2, 2), (0, 0))  # create a 2,2 grid and plot at the 0,0 location (top left)
plt.plot(x, np.cos(x))
plt.ylabel(r'$\cos(x)$')
plt.subplot2grid((2, 2), (0, 1))  # create a 2,2 grid and plot at te 0,1 location (top right)
plt.plot(x, np.sin(x))
plt.ylabel(r'$\sin(x)$')
plt.subplot2grid((2, 2), (1, 0), colspan=2)  # create a 2,2 grid and plot at the 1,0 (bottom left) but span 2 col
plt.plot(x, np.cos(x), label=r'$\cos(x)$')
plt.plot(x, np.sin(x), label=r'$\sin(x)$')
plt.ylabel(r'$\cos(x),\ \sin(x)$')
plt.xlabel('x')
plt.suptitle('Trigonometric functions')
plt.tight_layout()  # adjust the padding between plots at a fraction of the font size (1.08 def)
plt.show()
```

<span style="color: #E43F6F; font-size: 22px;"><strong>II. Scatter plots</strong></span>

The function <tt>scatter</tt> <span style="color: #AB68B2;"><i>plots only points</i></span> but, unlike the function <tt>plot</tt>, allows you to <span style="color: #AB68B2;"><i>change the size and colour of the points</i></span>, and even to change the symbol for each pair $(x_i,\, y_i)$. This allows you to plot points from different experiments using different symbols or colours. For more details, see <strong>§III.3.1</strong> of the lecture notes.

There are multiple ways of specifying colours.
<ul style="list-style: none">
  <li><span style="font-size:11px"> ▶</span> You can provide a single color, all points are then plotted with the same color (as for the function <tt>plot</tt>);</li>
  <li><span style="font-size:11px"> ▶</span> You can provide a list of color names (with the length of the list being equal to the number of data points);</li>
  <li><span style="font-size:11px"> ▶</span> You can provide a 2D array of floats between 0 and 1 of shape $(n,\, 3)$ (with $n$ the number of data points), each row then corresponds to the values of the three channels R, G and B;</li>
  <li><span style="font-size:11px"> ▶</span> You can provide a 1D array of floating values between 0 and 1 of size $n$ (with $n$ the number of data points) which will then be mapped on a colormap (the full list of colormaps is given there <a href=https://matplotlib.org/stable/tutorials/colors/colormaps.html>https://matplotlib.org/stable/tutorials/colors/colormaps.html</a>).</li>
</ul>

There are two ways of specifying sizes.
<ul style="list-style: none">
  <li><span style="font-size:11px"> ▶</span> You can provide a single float, all points are then plotted with the same size (as for the function <tt>plot</tt>);</li>
  <li><span style="font-size:11px"> ▶</span> You can provide a 1D array of size $n$ (with $n$ the number of data points) which corresponds to the size of each point.</li>
</ul>

The Mandelbrot set is a fractal defined as the set of points $c$ of the complex plane for which the sequence of complex numbers defined by the recurrence relation $z_{n+1} = z_n^2  + c$, with $z_0=0$, remains bounded. One can show that if the sequence of moduli of $z_n$ is strictly greater than 2 for a certain index $i_\mathrm{div}$ then this sequence is increasing from this index, and tends to infinity.

For a mesh of size $N \times M$ in the complex plane, in the limits $[−1.4,\, 1.4]$ on the complex axis and $[−2,\, 0.8]$ on the real axis, create an image which shows the index $i_\mathrm{div}$ from which the series diverges. This figure should be understandable to anyone.

Only one loop is allowed (20 iterations are sufficient for the figure).

```python

import numpy as np
import matplotlib.pyplot as plt

n = 1000
m = 1000

re = np.linspace(-2, 0.8, n)
img = np.linspace(-1.4, 1.4, m)

ii, jj = np.meshgrid(re, img)

c = ii + jj*1j

indexes = np.full_like(c, 100, dtype=int)
mask = np.full_like(c, False, dtype=bool)


z_prev = 0
n = 0

while n < 20:
    z = z_prev**2 + c
    z_prev = z

    indexes[(abs(z) > 2) & (~mask)] = n
    mask[abs(z) > 2] = True
    n += 1 


fig, ax = plt.subplots()

cmap = plt.get_cmap('plasma').copy()
cmap.set_over('white')
sc = ax.imshow(indexes, cmap=cmap, vmin=0, vmax=20, extent=[re.min(), re.max(), img.min(), img.max()], origin='lower')


ax.grid()
ax.set_xlabel("Re(y)")
ax.set_ylabel("Img(y)")
ax.grid(False)
cbar = fig.colorbar(sc, label=r"$i_{\text{div}}$")

plt.show()




```

A continued fraction is a mathematical expression that can be writen as: $\displaystyle q_0+\dfrac{1}{q_1+\dfrac{1}{q_2+\dfrac{1}{q_3+\dots}}}$.

This representation can be used to demonstrate the irrationality of certain numbers ($\pi$, $e$) when the continued fraction is infinite. In diophantine approximation, continued fractions provide excellent approximations of reals by rationals.

Let $x$ be any real number. It can be written as $x = q_0 + e_0$ where $q_0$ is an integer, the floor of $x$ (integer part or integral part, <i>i.e.,</i> $q_0 = \lfloor x\rfloor$), and $e_0$ is the remainder, so the fractional part ($e_0 = x − \lfloor x\rfloor$).

By construction $e_0<1$, so it can be written as $1/e_0 = q_1 + e_1 = \lfloor(1/e_0)\rfloor + e_1$ with $0 \leq e_1 < 1$. Similarly,
$1/e_1 = q_2 + e_2 = \lfloor(1/e_1)\rfloor + e_2$, etc. We then establish recurrence relations for $n \geq 0$:
$$\begin{cases}
q_{n+1}=\Bigl\lfloor\dfrac{1}{e_n}\Bigr\rfloor,\\
e_{n+1}=\dfrac{1}{e_n}-q_{n+1}=\dfrac{1}{e_n}-\Bigl\lfloor\dfrac{1}{e_n}\Bigr\rfloor.
\end{cases}$$
The sequence of incomplete quotients $q_n$, called the canonical representation, is given in square brackets: $[q_0,\,q_1,\,q_2,\dots,\, q_n]$.
For example $\dfrac{15 625}{6842}$ has $[2,\,3,\,1,\,1,\,9,\,1,\,1,\,48]$ as a canonical representation.

1) Write the function <tt>canonical_rep</tt> which calculates and returns the $n$ significant values of the sequence $q_n$ as a function of $x$ and $\epsilon$ given as arguments, where $\epsilon$ is the precision to be achieved, such that $e_n\leq \epsilon$. You can use the example given above to test your program. Apply it also to $x = \exp(1)$ with a precision of $\epsilon=10^{-5}$. Your function should also take a maximum number of iterations (by default equal to 20) as an input in case of irrational numbers.

2) Write the function <tt>cont_frac</tt> which calculates and returns the continued fraction approximation of $x$ given as arguments. The other arguments given to the function must be the precision $\epsilon$ and the maximum number of iterations for the calculation of the canonical representation.

<i>Hint:</i> The continued fraction approximation corresponds to the first element $\alpha_0$ of the sequence $\alpha_i$, defined by the following recurrence relation for decreasing indices: $\alpha_i = q_i + 1/\alpha_{i+1}$ for $i = 0,\dots, n-1$ and $\alpha_n = q_n$.

3) Plot the difference between the continued fraction and $x = \exp(1)$ as a function of the size of the canonical representation $n$ for $n=1,\dots,20$.

```python
import numpy as np
import matplotlib.pyplot as plt

def canonical_rep(x, epsilon=1e-5, maxit=20):

        q0 = int(x)
        e0 = x - int(x)

        e = 1 # random limit for epsilon comparison
        n = 0 # iteration's tracker (for maxit)

        q = [q0]

        while e > epsilon and n < maxit:

            q += [int(1/e0)]
            e = 1/e0 - int(1/e0)

            e0 = e
            n += 1

        return q
            
```

```python
def cont_frac(x, epsilon=1e-5, maxit=20):
    q_list = canonical_rep(x, epsilon, maxit)
    
    frac = 1/q_list.pop()
    q_list.reverse()
    q0 = q_list.pop()

    for q in q_list:
        frac = 1/(q + frac)
        
    frac += q0

    return frac
```
