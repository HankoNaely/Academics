<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>2) Copies of mutable/immutable objects </i></span>

There are two ways of <span style="color: #AB68B2;"><i>copying</i></span> an object in Python.

<ul style="list-style: none">
  <li><span style="font-size:11px"> ▶</span> <span style="color: #AB68B2;"><i>Shallow copy:</i></span> This creates a <span style="color: #AB68B2;"><i>new object</i></span>, but its elements are still references to the same objects. If the elements are mutable, changes inside them will affect both copies.</li>
  <li><span style="font-size:11px"> ▶</span> <span style="color: #AB68B2;"><i>Deep copy:</i></span> This creates a <span style="color: #AB68B2;"><i>new object</i></span>, and also recursively copies all objects inside it. Changes to the original object or its elements (even mutable ones) have no affect on the deep copy.</li>
</ul>

This is related to <span style="color: #AB68B2;"><i>memory allocation</i></span>, which can be probed using the function <tt>id</tt>.

<strong>Be careful that an assignment is not a copy</strong>. When you <span style="color: #AB68B2;"><i>assign</i></span> one variable to another (<i>e.g.,</i> <tt>b = a</tt>), both variables refer to the <span style="color: #AB68B2;"><i>same object</i></span> in memory. If the object is mutable, modifications through one variable will be visible through the other, because both variables share the same object. If the object is immutable, you cannot modify the object in place: any "modification" (like <tt>a = a + 1</tt>) will actually create a new object and rebind the name <tt>a</tt>, leaving <tt>b</tt> unchanged. For another explanation, you can consult: https://docs.python.org/3/faq/programming.html#why-did-changing-list-y-also-change-list-x

```python
a = 1.  # COMMENT REQUIRED
b = a  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))  # COMMENT REQUIRED
b += 1.
print('id of a =', id(a))
print('id of b =', id(b))  # COMMENT REQUIRED
a *= 2.
print('id of a =', id(a))
print('id of b =', id(b))
```

```python
a = ['a', 'b', 'c']  # COMMENT REQUIRED
b = a  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))
b[0] = 'aa'
print('id of a =', id(a))
print('id of b =', id(b))
print('a =', a)
print('b =', b)  # COMMENT REQUIRED
a[0] = 'a'
print('a =', a)
print('b =', b)
```

```python
a = ('a', 'b', 'c')  # COMMENT REQUIRED
b = a  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))  # COMMENT REQUIRED
a += tuple('d')
print('id of a =', id(a))
print('id of b =', id(b))  # COMMENT REQUIRED
print('a =', a)
print('b =', b)
```

```python
a = ['a', 'b', 'c']
b = list(a)  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))
a[1] = 'd'
print('a =', a)
print('b =', b)  # COMMENT REQUIRED
```

```python
a = [[1, 2], [3, 4]]
b = list(a)
print('id of a =', id(a))
print('id of b =', id(b))
print('a =', a)
print('b =', b)
a[0][0] = 0  # COMMENT REQUIRED
print('a =', a)
print('b =', b)  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))
```

```python
import copy  # We import the package copy to perform deep copies.
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))
print('a =', a)
print('b =', b)
a[0][0] = 0  # COMMENT REQUIRED
print('a =', a)
print('b =', b)  # COMMENT REQUIRED
print('id of a =', id(a))
print('id of b =', id(b))
```

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>3) Mutable/immutable arguments of functions/procedures</i></span>

In Python, when you pass a variable to a function parameter, the function <span style="color: #AB68B2;"><i>uses the same object as the one assigned to that variable</i></span>. So if the object is mutable, the function can change it, and the changes will be <span style="color: #AB68B2;"><i>visible outside the function</i></span>.

  <tr>

```python
def modify_value(r):
    r += 1.
    return r


my_float = 1.
print(my_float, id(my_float))

modify_value(my_float)  # COMMENT REQUIRED
print(my_float, id(my_float))

my_float = modify_value(my_float)  # COMMENT REQUIRED
print(my_float, id(my_float))
```

```python
def modify_list(tu):
    tu += tuple(['modified'])
    return tu


my_tuple = ('0', '0', '0')
modify_list(my_tuple)  # COMMENT REQUIRED
print(my_tuple, id(my_tuple))

my_new_tuple = modify_list(my_tuple)
print(my_tuple, id(my_tuple))  # COMMENT REQUIRED
print(my_new_tuple, id(my_new_tuple))  # COMMENT REQUIRED
```

```python
def modify_list(li):
    li[0] = 'modified'


my_list = 3 * [0.]
print(my_list, id(my_list))

modify_list(my_list)  # COMMENT REQUIRED
print(my_list, id(my_list))
```

```python
def modify_list(li):
    l_mod = li[:]  # COMMENT REQUIRED
    l_mod[0] = 'modified'
    return l_mod


my_list = 3 * [0.]
print(my_list, id(my_list))

my_new_list = modify_list(my_list)  # COMMENT REQUIRED
print(my_list, id(my_list))
print(my_new_list, id(my_new_list))
```
