<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>1) Basic types and assignment </i></span>

Like most languages, Python has a number of basic types. For more details, consult the lecture notes <strong>§I.2.1</strong>.

<ul style="list-style: none">
  <li><span style="font-size:11px"> ▶</span> Boolean (<tt>bool</tt>): the possible values are <tt>False</tt> or <tt>True</tt>;</li>

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>4) Sequences </i></span>

Python has variable types that are less common in scientific programming languages. These are called composite types. They are used for positionally ordered (<tt>list</tt> and <tt>tuple</tt>) or unordered (<tt>dict</tt> and <tt>set</tt>) <span style="color: #AB68B2;"><i>sequences</i></span>. For more details, consult the lecture notes <strong>§I.2.3</strong>.

<ul style="list-style: none">
  <li><span style="font-size:11px"> ▶</span> List (<tt>list</tt>): ordered collection of items separated by commas, all enclosed in square brackets;</li>
  <li><span style="font-size:11px"> ▶</span> Tuple (<tt>tuple</tt>): immutable list enclosed in brackets (contrary to lists, elements of a tuple cannot be modified);</li>
  <li><span style="font-size:11px"> ▶</span> Range (<tt>range</tt>): immutable sequence of ordered integers with <tt>range(start, end, step)</tt> creating a sequence from <tt>start</tt> (included) to <tt>end</tt> (excluded), by steps of <tt>step</tt>;</li>
  <li><span style="font-size:11px"> ▶</span> Dictionary (<tt>dict</tt>): sequences whose items are addressable with a key rather than an index;</li>
  <li><span style="font-size:11px"> ▶</span> Set (<tt>set</tt>): immutable unordered sequence of objects enclosed in angular brackets.</li>
</ul>

The concept of <span style="color: #AB68B2;"><i>(im)mutability</i></span> will be discussed more extensively in the next tutorial.

<span style="color: #E43F6F; font-size: 22px;"><strong>III. Conditional branching</strong></span>

<span style="color: #AB68B2;"><i>Conditional branching</i></span> is where a program decides whether to do something or not according to the result of some condition. Therefore, blocks of instructions must be defined that are executed or not depending on the test execution. See <strong>§I.3.1</strong> of the lecture notes for more details.

```python
year_of_birth = input('Enter your year of birth (for example 1994):')  # input() allows interaction with the user (value input).
print(type(year_of_birth))  # input() returns a string.
year = int(year_of_birth)
if year >= 1961 and year <= 1980:
    print('You belong to Generation X.')
elif year <= 1995:
    print('You belong to Generation Y (or Millenial Generation).')
elif year <= 2010:
    print('You belong to Generation Z (or iGen).')
else:
    print('You belong to Generation Alpha.')
```

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>1) <tt><span style="font-size: 18px;">for</span></tt> loops </i></span>

The <tt>for</tt> statement repeats a set of instructions by <span style="color: #AB68B2;"><i>iterating over a set of values stored in an iterable object</i></span> (lists, tuples or strings). Unlike many other languages, we do not give a <tt>min</tt>, <tt>max</tt> and <tt>step</tt>, but the behaviour of other languages (numerical iteration) can be simulated with the intrinsic function <tt>range()</tt> (see above). For more details, consult the lecture notes <strong>§I.3.2</strong>.

```python
for i in range(3):
    print(i)
```

```python
for i in range(3):
    if i < 2:
        print(i)
    else:
        print('Too large')
        break  # COMMENT REQUIRED
```

```python
print('for loop with else:')
for i in range(10):
    if i % 2 == 0:
        print(i, 'even')
    else:
        print(i, 'odd')

print('for loop with continue and without else:')
for i in range(10):
    if i % 2 == 0:
        print(i, 'even')
        continue
    print(i, 'odd')
```

```python
a = ["Marc", "est", "dans", "le", "jardin"]
for i in range(len(a)):
    description = "i vaut " + str(i) + " et c[" + str(i) + "] vaut " + str(a[i])
    print(description)
```

```python
a = ["Marc", "est", "dans", "le", "jardin"]
for aa in a:  # COMMENT REQUIRED
    print(aa)
```

```python
a = ["Marc", "est", "dans", "le", "jardin"]
for i, aa in enumerate(a):
    description = "i vaut " + str(i) + " et c[" + str(i) + "] vaut " + aa
    print(description)
```

```python
a = ["Marc", "est", "dans", "le", "jardin"]
b = enumerate(a)
print(b)  # COMMENT REQUIRED
print(type(b))
print(list(b))  # COMMENT REQUIRED
```

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>Exercise 6:</i></span>

Write a program that converts an integer number of seconds given at the beginning, into a number of days, hours, minutes and seconds (use the modulo operator: <tt>%</tt>).

For instance, 242 203 should return 2 days, 19 hours, 16 minutes, and 43 seconds.

```python
try:
    x = int(input("Enter an integer or consequences."))
except ValueError:
    print("That's not an integer you moron !")
else:
    m = x // 60 # Minutes in total
    s = x % 60 # Seconds that don't add up to 1m
    h = m // 60 # Hours in total
    m %= 60 # Minutes that don't add up to 1h
    d = h // 24 # Days in total
    h %= 24 # Hours that don't add up to a day

print(f"{d} days, {h} hours, {m} minutes, and {s} seconds.")
```

Write a program that displays the first 20 terms of the multiplication table by 7, indicating (with an asterisk) those that are multiples of 3.

Example: 7 14 21* 28 35 42* 49

```python
x = 7
table = []

for i in range(1, 21):
    table += [7*i]
    if table[i-1] % 3 == 0:
        table[i-1] = str(table[i-1]) + "*"
    else:
        table[i-1] = str(table[i-1])

print(" ".join(table))
```

<span style="color: #AB68B2; font-size: 18px; margin-left: 2em;"><i>Exercise 10:</i></span>

Write a program which asks for an integer between 1 and 3999 and displays it as a Roman numeral. Look on Wikipedia if you do not remember Roman numerals.

Your code should reproduce the following examples:

3 = III

4 = IV

8 = VIII

9 = IX

10 = X

39 = XXXIX

42 = XLII

509 = DIX

666 = DCLXVI

1992 = MCMXCII

```python
try:
    x = int(input("Enter an integer between 1 and 3999 or consequences (bad ones)."))
except ValueError:
    print("That's no integer you flawed human !")
if x < 1 or x > 3999:
    raise Exception("That's not between 1 and 3999 you moron !")

# Feeling the beginning of the string with zeroes so that the conversion doesn't mix different order of magnitudes.
if x < 10:
    filled_number = 3*"0" + str(x)
elif x < 100:
    filled_number = 2*"0" + str(x)
elif x < 1000:
    filled_number = "0" + str(x)
else:
    filled_number = str(x)


number = enumerate(filled_number, start=1) 

units = {
    "0": "",
    "1": "I", 
    "2": "II",
    "3": "III",
    "4": "IV",
    "5": "V",
    "6": "VI",
    "7": "VII",
    "8": "VIII",
    "9": "IX"
}

tens = {
    "0": "",
    "1": "X", 
    "2": "XX",
    "3": "XXX",
    "4": "XL",
    "5": "L",
    "6": "LX",
    "7": "LXX",
    "8": "LXXX",
    "9": "XC"
}

hundreds = {
    "0": "",
    "1": "C", 
    "2": "CC",
    "3": "CCC",
    "4": "CD",
    "5": "D",
    "6": "DC",
    "7": "DCC",
    "8": "DCCC",
    "9": "CM"
}

thousands = {
    "0": "",
    "1": "M",
    "2": "MM",
    "3": "MMM"
}

roman_conversion = ''

for i, n in number:
    if i == 1:
        reference = thousands
    elif i == 2:
        reference = hundreds
    elif i == 3:
        reference = tens
    elif i == 4:
        reference = units

    roman_conversion += reference[n]

print(f"{x} = " + roman_conversion)


```
